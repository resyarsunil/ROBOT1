# Autonomous Navigation allows the robot to understand their location and travel without human intervention.


Occupancy grid: The space is divided into a grid of cells. Each cell stores a value indicating whether it is empty, occupied or unknown.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
grid_size = (20,20)

grid = np.zeros(grid_size)
grid[5:15, 5:15] = 1            #obstacle in x and y
plt.figure(figsize=(6,6))
plt.imshow(grid.T, origin='lower', cmap='gray_r')
plt.title('2D occupancy grid')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.xticks(range(0, grid_size[0]+1, 1))
plt.yticks(range(0, grid_size[1]+1, 1)) 
plt.show()
```
